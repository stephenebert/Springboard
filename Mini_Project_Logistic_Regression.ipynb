Let's make a correlation heatmap

```python
sns.heatmap(features.corr(), cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

This correlation heatmap shows the linear relationships between the features in the breast cancer public dataset. Dark red regions indicate strong positive correlations, while dark blue regions indicate strong negative correlations.

One notable pattern is that several features related to tumor size and shape such as mean radius, mean perimeter, and mean area—are strongly positively correlated with one another. This suggests that they may be capturing overlapping information, which could introduce multicollinearity into statistical models.


Furthermore, the "worst" versions of features like worst radius, worst perimeter, and worst concavity also show high intercorrelation. In contrast, features like error terms (e.g. smoothness error, symmetry error etc) tend to have weaker or more variable correlations.

The multicollinearity assumption appears to be **broken**, as shown in the correlation heatmap of features above. Multicollinearity occurs when independent variables (features) are strongly correlated with each other. This can lead to unstable coefficient estimates and increased variance in model predictions. As a result, the model becomes more sensitive to noise, increasing the risk of overfitting. Therefore, we need to be cautious and consider techniques to address multicollinearity before training the model.

# **Part 2: Building a Logistic Regression Model with scikit-learn**

**1. Use train_test_split to create a training and test sets for the data.**

```python
from sklearn.model_selection import train_test_split
#x_train, x_test, y_train, y_test notice the naming and ordering is "train-test" so we can remember better this train -> test, train -> test
```

```python
X = features
y = target
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=30) # the usual 80-20 split (80% train & 20% split where we have a seed for reproducibility)
```

Now we have our training and test data established, we are ready to make our logistic model by fitting our training data

**2. Use LogisticRegression to train a model on the training data. Make sure you understand the inputs to the model. Try using the "liblinear" solver here.**

```python
from sklearn.linear_model import LogisticRegression
logistic_model = LogisticRegression(solver = 'liblinear')
logistic_model.fit(X_train, y_train)
```

# **Part 3: Evaluating Logistic Model with a Confusion Matrix and Classification Report**

**1. Use your model to make predictions on the test data.**

To make a prediction, we just plug in our test data "X_test" into our logistic model "logistic_model" and store that in "y_prediction"

```python
y_prediction = logistic_model.predict(X_test)
```

**2. Generate a confusion matrix with the test results. How many false positives and false negatives did the model predict?**

```python
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report
confusion_matrix(y_test, y_prediction)
```

```python
print("Confusion Matrix:\n", confusion_matrix(y_test, y_prediction))
```

```python
report = classification_report(y_test, y_prediction)
print("Classification Report:\n", report)
```

```python
#we can also visualize the confusion matrix through a heatmap
sns.heatmap(confusion_matrix(y_test, y_prediction), annot=True, fmt='d', cmap='coolwarm')
plt.xlabel("Predicted Label")
plt.ylabel("True Label")
plt.title("Confusion Matrix")
plt.show()
```

After training the logistic model, I checked how well it was doing by looking at the confusion matrix (tells me how many predictions the model got right and where it messed up). Out of all the test cases, it correctly predicted most of them, with just a few false alarms and a few it missed. I also looked at a report that gave me numbers like precision and recall, which are ways to measure how careful and thorough the model is. The scores were pretty high across the board, meaning the model is doing a solid job! The heatmap helped me visualize this more clearly. Overall, the model is working well.

# **Part 4: Coefficients & Visualization**

**1. Extract the model coefficients from your trained model.**

```python
coefficients = logistic_model.coef_[0]  # shape: (n_features,)
coefficients
```

**2. Normalize the coefficients by the standard deviation of each feature in the training data.**

```python
standard_deviations = X_train.std(axis=0)
normalized_coefficients = coefficients / standard_deviations
normalized_coefficients
```

**3. Sort feature names and coefficients by absolute value of coefficients.**

```python
feature_labels = data.feature_names
norm_coefs = normalized_coefficients

coef_df = pd.DataFrame({
    'feature': feature_labels,
    'norm_coef': norm_coefs
})

coef_df['abs_coef'] = coef_df['norm_coef'].abs()

# sort the features by absolute coeff
coef_df_sorted = coef_df.sort_values(by='abs_coef', ascending=False)

# display sorted coeffs
coef_df_sorted.reset_index(drop=True, inplace=True)
coef_df_sorted
```

**4. Visualize the feature importances by creating a horizontal bar chart using e.g. barh. Based on magnitude, what appears to be the most important predictor of cancer in this dataset?**

```python
plt.figure(figsize=(10, 8))
plt.barh(coef_df_sorted['feature'], coef_df_sorted['norm_coef'], color='black')
plt.xlabel('Normalized Coefficient')
plt.title('Feature Importance w/ Logistic Regression')
plt.show()
```

"**Worst smoothness**" is the most important predictor of cancer in this dataset. It makes sense as the most important predictor because malignant tumors often have highly irregular, jagged edges, which this feature captures at its most extreme.
