# Paths and settings

```python
PARQUET = r"C:/Users/steph/OneDrive/Desktop/data/metadata.parquet"
PRESET = "cpu-fast"          
BATCH_OVERRIDE =    None        
MAX_SAMPLES_OVERRIDE = 2000      
MODEL_OVERRIDE       = None    # e.g. "RN50"
PRETRAINED_OVERRIDE  = None    # e.g. "openai"
```

# Imports and preset

```python
import json, time, datetime, sys
from pathlib import Path

import numpy as np
import torch
from torch.utils.data import DataLoader, Subset
from tqdm import tqdm
import open_clip
from open_clip import tokenize
import sys
sys.path.append(r"C:\Users\steph\OneDrive\Desktop\cross-modal-retrieval\src")

from retrieval import RetrievalDataset

repo_root = Path.cwd()
if not (repo_root / "src").exists():
    repo_root = repo_root.parent
sys.path.append(str(repo_root / "src"))


PRESETS = {
    "gpu": {
        "device": "cuda",
        "model": "ViT-B-32",
        "pretrained": "laion2b_s34b_b79k",
        "batch": 128,
        "max_samples": None,
    },
    "cpu-fast": {
        "device": "cpu",
        "model": "RN50",
        "pretrained": "openai",
        "batch": 64,
        "max_samples": 50_000,
    },
    "cpu-tiny": {
        "device": "cpu",
        "model": "RN50",
        "pretrained": "openai",
        "batch": 64,
        "max_samples": 10_000,
    },
}

cfg = PRESETS[PRESET].copy()
if BATCH_OVERRIDE is not None:
    cfg["batch"] = BATCH_OVERRIDE
if MAX_SAMPLES_OVERRIDE is not None:
    cfg["max_samples"] = MAX_SAMPLES_OVERRIDE
if MODEL_OVERRIDE is not None:
    cfg["model"] = MODEL_OVERRIDE
if PRETRAINED_OVERRIDE is not None:
    cfg["pretrained"] = PRETRAINED_OVERRIDE

device = cfg["device"]
if device == "cuda" and not torch.cuda.is_available():
    print("CUDA not available - switching to CPU")
    device = "cpu"
```

# dataset and model

```python
ts = datetime.datetime.now().strftime("%Y%m%d_%H%M%S")
exp_dir = Path("experiments") / f"{cfg['model']}_{ts}"
exp_dir.mkdir(parents=True, exist_ok=True)
print("Experiment dir:", exp_dir.resolve())

# dataset
full_ds = RetrievalDataset(PARQUET)
if cfg["max_samples"]:
    rand_idx = torch.randperm(len(full_ds))[: cfg["max_samples"]].tolist()
    dataset = Subset(full_ds, rand_idx)         
else:
    dataset = full_ds

loader = DataLoader(
    dataset,
    batch_size=cfg["batch"],
    shuffle=False,
    num_workers=4,
    pin_memory=(device == "cuda"),
)
print(f"Dataset size: {len(dataset):,}  |  batch {cfg['batch']}")

# model
model, _, _ = open_clip.create_model_and_transforms(
    cfg["model"], pretrained=cfg["pretrained"], device=device
)
model.eval()
print(f"Model {cfg['model']} ({cfg['pretrained']}) on {device}")
```

#  Embedding loop

```python
start = time.time()
img_parts, txt_parts, ids = [], [], []

with torch.no_grad():
    for batch in tqdm(loader, desc="Embedding", mininterval=1):
        imgs = batch["image"].to(device, non_blocking=True)
        caps = tokenize(batch["caption"]).to(device, non_blocking=True)

        ie = model.encode_image(imgs)
        ie = ie / ie.norm(dim=-1, keepdim=True)

        te = model.encode_text(caps)
        te = te / te.norm(dim=-1, keepdim=True)

        img_parts.append(ie.cpu())
        txt_parts.append(te.cpu())
        ids.extend(batch["id"])

img_arr = torch.cat(img_parts).numpy().astype("float16")
txt_arr = torch.cat(txt_parts).numpy().astype("float16")

np.save(exp_dir / "img_embs.npy", img_arr)
np.save(exp_dir / "txt_embs.npy", txt_arr)
json.dump(ids, open(exp_dir / "ids.json", "w"))
print("Embeddings saved")
```

# Metrics and config snapshot

```python
sim = txt_arr @ img_arr.T

def r_at_k(mat, k):
    topk = np.argpartition(-mat, k - 1, axis=1)[:, :k]
    return float(np.mean(np.any(topk == np.arange(len(mat))[:, None], axis=1)))

metrics = {f"R@{k}": round(r_at_k(sim, k) * 100, 2) for k in (1, 5, 10)}
ranks = np.argsort(-sim, axis=1)
med = [int(np.where(ranks[i] == i)[0][0] + 1) for i in range(len(sim))]
metrics["median_rank"] = int(np.median(med))
json.dump(metrics, open(exp_dir / "metrics.json", "w"), indent=2)
print("Metrics:", metrics)

cfg_out = dict(cfg, parquet=str(Path(PARQUET).resolve()), device=device)
json.dump(cfg_out, open(exp_dir / "config.json", "w"), indent=2)
elapsed = round((time.time() - start) / 60, 1)
print("Finished in", elapsed, "minutes")
```

```python
import os
import json
from pathlib import Path
import matplotlib.pyplot as plt
import pandas as pd

# Step 1: Load experiment metadata and metrics
EXP_ROOT = Path("experiments")
data = []

for run in sorted(EXP_ROOT.iterdir()):
    try:
        metrics_path = run / "metrics.json"
        config_path = run / "config.json"
        if not metrics_path.exists() or not config_path.exists():
            continue

        metrics = json.loads(metrics_path.read_text())
        config = json.loads(config_path.read_text())
        samples = config.get("max_samples") or len(json.loads((run / "ids.json").read_text()))

        row = {
            "run": run.name,
            "samples": samples,
            "Recall@1": metrics.get("R@1"),
            "Recall@5": metrics.get("R@5"),
            "Recall@10": metrics.get("R@10"),
        }
        data.append(row)
    except Exception as e:
        print(f"Skipping {run.name}: {e}")

# Step 2: Create DataFrame and sort by sample size
df = pd.DataFrame(data)
df = df.dropna().sort_values("samples")

# Step 3: Plotting
plt.figure(figsize=(10, 6))
plt.plot(df["samples"], df["Recall@1"], marker='o', label="Recall@1")
plt.plot(df["samples"], df["Recall@5"], marker='o', label="Recall@5")
plt.plot(df["samples"], df["Recall@10"], marker='o', label="Recall@10")
plt.xlabel("Max Samples Used")
plt.ylabel("Accuracy (%)")
plt.title("Retrieval Accuracy vs. Dataset Size")
plt.legend()
plt.grid(True)
plt.savefig("scaling_plot.png")
plt.show()

df
```
